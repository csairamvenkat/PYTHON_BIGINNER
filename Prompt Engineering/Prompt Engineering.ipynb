```python
# Using Input Function to Create a Prompt
def create_prompt():
 topic=input("Enter topic for story")
 tone=input("Entrr tone for story eg : Serious,comedy etc")
 
 prompt=f"Write a {tone} story about {topic}."
 return prompt


user_prompt=create_prompt()
print("Your prompt is:",user_prompt)
```

```python
# Using Lits
topics=["AI","ISRO","Indian Histrory"]
tones=["Serious","Comedy","Suspense"]

prompts=[]
for topic in topics:
    for tone in tones:
        prompt=f"Write a {tone} story about {topic}."
        prompts.append(prompt)
        
print("Generated prompts:")
for p in prompts:
    print(p)
```

```python
#Using Templates
def create_summary_prompt(text):
    prompt_template="Summerize the following Text :{text}"
    return prompt_template.format(text=text)

input_text=input("Enter text to summarize:")
summary_prompt=create_summary_prompt(input_text)
print("Your summary prompt is:", summary_prompt)
```
