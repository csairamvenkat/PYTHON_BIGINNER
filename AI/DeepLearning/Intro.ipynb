# Forward Propogations : 
Input to hidden to output

# Backward Propogation :
output to hidden to input

 # Epoch : 1 Iteration is called Epoch-->forward and backward is 1 iteration
 Epoch increases accuracy (low cost function and high accuracy)

# Fine Tuning of errors :
 -->Fwd Propogation we get few weights abut finally we get wrong prediction. Reduce loss function and do backward propogation,this gives new weights==>completes 1 epoch
 -->Start with new weights from backward propogation and do a fwd propogation-->accuracy increasd-->do backward propogation and get new weights (Epoch 2)
 -->Similarly do multiple epochs to get better accuracy
