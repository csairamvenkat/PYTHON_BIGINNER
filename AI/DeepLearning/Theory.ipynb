# Transfer Learning :
Keras-->It has lot of pretrained mopdels
Pretrained Data sets are called pre trained weights.

Transfer Learning is a machine learning technique where a model developed for one task is reused as the starting point for a model on a second, related task. It’s especially powerful in deep learning where models require a lot of data and computational resources to train.
📌 Key Idea:
Instead of training a model from scratch, you take a pre-trained model (trained on a large dataset like ImageNet or a language corpus) and fine-tune it for your specific task.
🧠 How It Works:
Pretrain: A model is trained on a large dataset to learn general patterns (e.g., edge detection in images or syntax in text).

Transfer:

Feature extraction: Use the pretrained model as a fixed feature extractor — only train the last layer(s).

Fine-tuning: Unfreeze some of the pretrained layers and train them along with new layers to adapt to your task.

| Domain | Pretrained Models         | Example Tasks                            |
| ------ | ------------------------- | ---------------------------------------- |
| Vision | ResNet, VGG, EfficientNet | Image classification, object detection   |
| NLP    | BERT, GPT, RoBERTa, T5    | Sentiment analysis, translation          |
| Audio  | Wav2Vec, DeepSpeech       | Speech recognition, audio classification |
