# Tokenization

-->Breaking complex sentence to words.
Tokenization is a fundamental step in Natural Language Processing (NLP). It involves breaking down text into smaller units called tokens. These tokens can be:

Words (word-level tokenization)

Characters (character-level tokenization)

Subwords (used in models like BERT or GPT)

Example of Word Tokenization:
Input: "Natural Language Processing is fun!"
Output: ["Natural", "Language", "Processing", "is", "fun", "!"]

Types of Tokenization:
Word Tokenization – Splits text by spaces and punctuation.

Character Tokenization – Splits text into individual characters.

Subword Tokenization – Splits words into smaller units to handle rare or unknown words better (e.g., ["un", "##break", "##able"] in BERT).

Sentence Tokenization – Splits a text into sentences.

Why Tokenization is Important:
It prepares text for further processing like POS tagging, parsing, or feeding into machine learning models.

It helps machines understand the structure of the language.

# Stemming :its of 3 types
Normalize words to base or root form

  Porter Stemmer
  Lancaster Stemmer
  Snow ball stemmer

| Feature                | **Porter Stemmer**                         | **Lancaster Stemmer**                         |
|------------------------|--------------------------------------------|-----------------------------------------------|
| **Aggressiveness**     | **Moderate** – keeps more of the word intact | **Highly aggressive** – may over-stem words   |
| **Speed**              | Slower                                     | Faster (due to simpler rules)                 |
| **Stemming Rules**     | Complex and context-sensitive              | Uses a lookup table and simple rules          |
| **Accuracy**           | Generally more accurate for information retrieval | Less accurate; can stem words too much      |
| **Example**            | `"running"` → `"run"`                      | `"running"` → `"run"` (same)<br>`"maximum"` → `"maximum"` | `"maximum"` → `"max"` |
| **Overstemming Risk** | Lower                                       | Higher                                        |
| **Use Case**           | When precision is needed                   | When performance or simplicity is preferred   |

| Feature            | Porter Stemmer | Lancaster Stemmer | Snowball Stemmer |
|--------------------|----------------|-------------------|------------------|
| Aggressiveness     | Medium         | High              | Low to Medium    |
| Accuracy           | Medium         | Low               | High             |
| Speed              | Moderate       | Fast              | Moderate         |
| Readability        | Good           | Poor              | Good             |
| Multilingual       | ❌             | ❌                | ✅               |

# Lametization :
It gives complete words
 text preprocessing step in Natural Language Processing (NLP). Lemmatization reduces a word to its base or dictionary form (lemma), taking into account the context (like part of speech).

Example:
"running" → "run"

"better" → "good"

This is different from stemming, which just chops off word endings without understanding the context:

"running" → "runn" (by a stemmer)

Named Entity Recognition (NER) is a key task in Natural Language Processing (NLP) that involves identifying and classifying named entities in text into predefined categories such as:
| Category | Description                      | Example                    |
|----------|----------------------------------|----------------------------|
| PERSON   | Names of people                  | *Elon Musk*, *Marie Curie* |
| ORG      | Organizations                    | *Google*, *UNICEF*         |
| GPE      | Countries, cities, states        | *India*, *New York*        |
| DATE     | Dates and times                  | *April 30, 2025*, *Monday* |
| TIME     | Specific times                   | *3 PM*, *noon*             |
| MONEY    | Monetary values                  | *$500*, *€1 million*       |
| LOC      | Locations (not geopolitical)     | *Himalayas*, *Sahara*      |
| PRODUCT  | Products                         | *iPhone*, *Coca-Cola*      |
| EVENT    | Named events                     | *World War II*, *Olympics* |

# Chunking

In NLP, chunks refer to shallow parsing or grouping words into meaningful phrases, such as noun phrases (NPs) or verb phrases (VPs). This is known as chunking—an intermediate step between POS tagging and full syntactic parsing.

🧱 What is Chunking?
Chunking = Grouping adjacent tokens (words) into phrases

Usually done using POS tags

Example:

Sentence: "The quick brown fox jumps over the lazy dog"
Noun Phrase Chunk: The quick brown fox, the lazy dog

| Chunk Type | Description                   | Example                          |
|------------|-------------------------------|----------------------------------|
| NP         | Noun Phrase                   | *The red car*                   |
| VP         | Verb Phrase                   | *is running fast*               |
| PP         | Prepositional Phrase          | *in the park*                   |


| Aspect     | Chunking                             | NER                                 |
|------------|--------------------------------------|-------------------------------------|
| Focus      | Grammatical structure (phrases)      | Semantic meaning (entities)         |
| Example    | "The quick brown fox" (NP)           | "New York" (GPE)                    |
| Based on   | POS patterns                         | Pretrained models / rules           |

# Natural Language Generation

Natural Language Generation (NLG) is a subfield of Natural Language Processing (NLP) focused on automatically producing human-like text from structured data or machine representations.

🧠 What is NLG?
NLG takes input like structured data, text summaries, or knowledge graphs, and generates fluent, coherent natural language.

It’s the reverse of Natural Language Understanding (NLU).

🔄 NLG Workflow (Pipeline)

Content Determination: What information to convey.

Text Structuring: Organizing the selected content logically.

Sentence Planning:

Choose words

Set tone and sentence structure

Surface Realization: Generate grammatically correct sentences.

(Optional) Post-processing: Adjust fluency or style.

| Application            | Example                                                              |
|------------------------|----------------------------------------------------------------------|
| Chatbots               | "Hello! How can I assist you today?"                                 |
| Report generation      | "Sales increased by 15% in Q1 2025 compared to last year."           |
| Image/Video captions   | "A dog jumping over a hurdle in a field."                            |
| Summarization          | Creating concise summaries from long documents.                      |
| Code generation        | Converting instructions to source code.                              |
