```python
# Gathering the spans of the tokenized strings.
from nltk.tokenize import TreebankWordTokenizer

tokenizer = TreebankWordTokenizer()

s = '''Good muffins cost $3.88\nin New (York).  Please (buy) me\ntwo of them.\n(Thanks).'''

expected_spans = [
    (0, 4), (5, 12), (13, 17), (18, 19), (19, 23),
    (24, 26), (27, 30), (31, 32), (32, 36), (36, 37), (37, 38),
    (40, 46), (47, 48), (48, 51), (51, 52), (53, 55), (56, 59),
    (60, 62), (63, 68), (69, 70), (70, 76), (76, 77), (77, 78)
]

actual_spans = list(tokenizer.span_tokenize(s))
print("Spans match:", actual_spans == expected_spans)

expected_tokens = [
    'Good', 'muffins', 'cost', '$', '3.88', 'in',
    'New', '(', 'York', ')', '.', 'Please', '(', 'buy', ')',
    'me', 'two', 'of', 'them.', '(', 'Thanks', ')', '.'
]

actual_tokens = [s[start:end] for start, end in actual_spans]
print("Tokens match:", actual_tokens == expected_tokens)
```

```python
# Regression Tests: Regexp Tokenizer
from nltk.tokenize import regexp_tokenize
s = ("Good muffins cost $3.88\nin New York.  Please buy me\n"
      "two of them.\n\nThanks.")
s2 = ("Alas, it has not rained today. When, do you think, "
       "will it rain again?")
s3 = ("<p>Although this is <b>not</b> the case here, we must "
       "not relax our vigilance!</p>")

# Discards words and gives all punctuaions
print('Gaps= False',regexp_tokenize(s2, r'[,\.\?!"]\s*', gaps=False))

# Discards Punctuation and gives all words
print('Gaps= True ',regexp_tokenize(s2, r'[,\.\?!"]\s*', gaps=True))

```

```python
# Take care to avoid using capturing groups:
print(regexp_tokenize(s3, r'</?[bp]>', gaps=False))
print(regexp_tokenize(s3, r'</?(?:b|p)>', gaps=False))
print(regexp_tokenize(s3, r'</?(?:b|p)>', gaps=True))
print(regexp_tokenize(s2, r'[,\.\?!"]\s*', gaps=False))
print(regexp_tokenize(s2, r'[,\.\?!"]\s*', gaps=True))

```

```python
# Take care to avoid using capturing groups:
print('Take care to avoid using capturing groups:')

print(regexp_tokenize(s3, r'</?[bp]>', gaps=False))
print(regexp_tokenize(s3, r'</?(?:b|p)>', gaps=False))
print(regexp_tokenize(s3, r'</?(?:b|p)>', gaps=True))
```

```python
# Named groups are capturing groups, and confuse the tokenizer:
print(regexp_tokenize(s3, r'</?(?P<named>b|p)>', gaps=False))
print(regexp_tokenize(s3, r'</?(?P<named>b|p)>', gaps=True))
```

```python
# Make sure that nested groups don’t confuse the tokenizer:
print(regexp_tokenize(s2, r'(?:h|r|l)a(?:s|(?:i|n0))', gaps=False))
print(regexp_tokenize(s2, r'(?:h|r|l)a(?:s|(?:i|n0))', gaps=True))
```

```python
# Back-references require capturing groups, and these are not supported:
print(regexp_tokenize("aabbbcccc", r'(.)\1'))
```
