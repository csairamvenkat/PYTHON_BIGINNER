```python
# Stop Words

import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Download stopwords once
nltk.download('stopwords')
nltk.download('punkt')

text = "This is a sample sentence showing how to remove stop words."

# Tokenize the text
words = word_tokenize(text)

# Remove stopwords
filtered_words = [word for word in words if word.lower() not in stopwords.words('english')]

print(filtered_words)
```

```python
# Parts of Speech
import spacy

nlp = spacy.load("en_core_web_sm")
doc = nlp("The quick brown fox jumps over the lazy dog")

for token in doc:
    print(f"{token.text} -> {token.pos_} ({token.tag_})")
```

```python
# Named Entity Recognization (NER)
import spacy

# Load the English NLP model
nlp = spacy.load("en_core_web_sm")

# Input text
text = "Apple is looking at buying U.K. startup for $1 billion on January 1st."

# Process the text
doc = nlp(text)

# Extract named entities
for ent in doc.ents:
    print(f"{ent.text} -> {ent.label_}")
```

```python
# NLG
# Chunking 
import nltk

# Define the grammar for noun phrases
grammar = "NP: {<DT>?<JJ>*<NN>}"

# Create a chunk parser
chunk_parser = nltk.RegexpParser(grammar)

# Tokenize and tag
sentence = "The quick brown fox jumps over the lazy dog"
tokens = nltk.word_tokenize(sentence)
tagged = nltk.pos_tag(tokens)

# Parse chunks
chunk_tree = chunk_parser.parse(tagged)

# Visualize chunks
chunk_tree.draw()  # Opens a window with a tree diagram

# Or print noun phrases
for subtree in chunk_tree.subtrees():
    if subtree.label() == 'NP':
        print("Noun Phrase:", " ".join(word for word, tag in subtree))
```
