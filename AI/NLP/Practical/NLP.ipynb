```python
import os
import nltk
# nltk.download()
```

```python
AI = '''Artificial Intelligence refers to the intelligence of machines. This is in contrast to the natural intelligence of
humans and animals. With Artificial Intelligence, machines perform functions such as learning, planning, reasoning and
problem-solving. Most noteworthy, Artificial Intelligence is the simulation of human intelligence by machines.
It is probably the fastest-growing development in the World of technology and innovation. Furthermore, many experts believe
AI could solve major challenges and crisis situations.'''

AI 
```

```python
type(AI)
```

```python
# Returns the Words
from nltk.tokenize import word_tokenize
AI_tokens=word_tokenize(AI)
print(AI_tokens)
print(len(AI_tokens))
```

```python
AI
```

```python
# Returns the Sentences
from nltk.tokenize import sent_tokenize
AI_sent=sent_tokenize(AI)
print(AI_sent)
print(len(AI_sent))
```

```python
# Returns the para
from nltk.tokenize import blankline_tokenize
AI_blank=blankline_tokenize(AI)
print(AI_blank)
print(len(AI_blank))
```

```python
# Returns the para
AI2='''Artificial Intelligence (AI) is transforming industries across the globe. From healthcare and finance to transportation and education, AI-powered systems are helping improve efficiency and decision-making. Natural Language Processing, a subfield of AI, enables machines to understand and respond to human language, making interactions with technology more intuitive.

Despite the rapid advancements, challenges still remain. Ethical concerns, data privacy, and algorithmic bias are just a few of the issues being debated. As researchers and developers continue to innovate, it is essential to ensure that these technologies are built responsibly and inclusively.'''
from nltk.tokenize import blankline_tokenize
AI2_blank=blankline_tokenize(AI2)
print(AI2_blank)
print(len(AI2_blank))
```

```python
# Removes punctuations
from nltk.tokenize import WhitespaceTokenizer
AI_whitespace=WhitespaceTokenizer().tokenize(AI)
print(AI_whitespace)
print(len(AI_whitespace))
```

```python
# Breaks all symobols ,numerics etc into each token
s='Good Apple cost $3.18 in hyd.Please but 2 of them.Thanks'
print(s)
from nltk.tokenize import wordpunct_tokenize
print(wordpunct_tokenize(s))
print(len(wordpunct_tokenize(s)))
```

```python
# Breaks all symobols ,numerics etc into each token
w_p=wordpunct_tokenize(AI)
print(w_p)
print(len(w_p))
```

```python
from nltk.util import bigrams,trigrams,ngrams
string = 'hello the best and most beautifull thing in the world cannot be seen or even touched,they must be felt with heart'
quotes_tokens = nltk.word_tokenize(string)
print(quotes_tokens)
print(len(quotes_tokens))
print('bigrams',list(bigrams(quotes_tokens)))
print('trigrams',list(trigrams(quotes_tokens)))
print('4-grams:', list(ngrams(quotes_tokens, 4)))  #ngrams
```

```python
# Stemming  : Normalize words to base or root form
# Porter Stemmer
from nltk.stem import PorterStemmer
pst=PorterStemmer()
print(pst.stem('Affection'))
print(pst.stem('Playing'))
print(pst.stem('Sing'))
print(pst.stem('max'))
```

```python
wordstostem=["Give","Giving","Thought",'Cricketing']
for  w in wordstostem:
  print(w, " : ",pst.stem(w))
```

```python
# Stemming  : Normalize words to base or root form
# Lancaster  Stemmer
from nltk.stem import LancasterStemmer
lst=LancasterStemmer()
print(lst.stem('Affection'))
print(lst.stem('Playing'))
print(lst.stem('Sing'))
print(lst.stem('max'))


wordstostem=["Give","Giving","Thought",'Cricketing']
for  w in wordstostem:
  print(w, " : ",lst.stem(w))
```

```python
# Stemming  : Normalize words to base or root form
# SnowBall  Stemmer
from nltk.stem import SnowballStemmer
sst=SnowballStemmer('english')
print(sst.stem('Affection'))
print(sst.stem('Playing'))
print(sst.stem('Sing'))
print(sst.stem('max'))


wordstostem=["Give","Giving","Thought",'Cricketing']
for  w in wordstostem:
  print(w, " : ",sst.stem(w))
```
