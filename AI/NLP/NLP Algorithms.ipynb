# Bag OF Words  :
 🧠 What is Bag of Words?
Bag of Words represents text data (like a sentence or document) as a set of words and their frequencies, ignoring grammar and word order but keeping multiplicity (how often a word appears).

🧮 Example:
Imagine you have these 2 sentences:

Sentence 1: "I love data"

Sentence 2: "I love AI and data"

Vocabulary: ["I", "love", "data", "AI", "and"]

| Word     | S1  | S2  |
|----------|-----|-----|
| I        | 1   | 1   |
| love     | 1   | 1   |
| data     | 1   | 1   |
| AI       | 0   | 1   |
| and      | 0   | 1   |






# 🧠 Word Embeddings in NLP – Explained Simply
Embeddings in NLP are a way to convert words into numerical vectors such that their semantic meaning is preserved. Unlike Bag of Words or TF-IDF, which are sparse and based on frequency, embeddings are dense, real-valued vectors that capture contextual similarity.

# TFIDF(Term Frequency *  Inverse Document Frequency)
# tf=(Number of times term t appears in a document/No. OF  terms  in document)
🧪 Example
Document: "AI is the future. AI will shape the future."

First, clean and tokenize:
["ai", "is", "the", "future", "ai", "will", "shape", "the", "future"]

| Term    | Count |
|---------|-------|
| ai      | 2     |
| is      | 1     |
| the     | 2     |
| future  | 2     |
| will    | 1     |
| shape   | 1     |

✅ Step 2: Total words in document = 9
✅ Step 3: Apply TF formula
| Term    | Frequency Count | TF = Count / 9 |
|---------|------------------|----------------|
| ai      | 2                | 2/9 ≈ 0.222     |
| is      | 1                | 1/9 ≈ 0.111     |
| the     | 2                | 2/9 ≈ 0.222     |
| future  | 2                | 2/9 ≈ 0.222     |
| will    | 1                | 1/9 ≈ 0.111     |
| shape   | 1                | 1/9 ≈ 0.111     |


# IDF=log( total number of documents/number of documents containing term t)


IDF Explanation :
| Document ID | Text                                 |
|-------------|--------------------------------------|
| Doc 1        | "AI is the future of the world"      |
| Doc 2        | "AI and data science are powerful"   |
| Doc 3        | "The world of science is expanding"  |

 ✅ Step 1: Document Frequency of each word

| Term        | Appears In Documents | Document Frequency (df) |
|-------------|----------------------|--------------------------|
| ai          | Doc1, Doc2           | 2                        |
| is          | Doc1, Doc3           | 2                        |
| the         | Doc1, Doc3           | 2                        |
| future      | Doc1                 | 1                        |
| world       | Doc1, Doc3           | 2                        |
| data        | Doc2                 | 1                        |
| science     | Doc2, Doc3           | 2                        |
| are         | Doc2                 | 1                        |
| powerful    | Doc2                 | 1                        |
| expanding   | Doc3                 | 1                        |
| of          | Doc1, Doc3           | 2                        |

✅ Step 2: Calculate IDF
Let total documents 
𝑁=3

| Term      | df(t) | IDF = log(N / (1 + df(t)))     | IDF Value (rounded) |
|-----------|-------|-----------------------------|----------------------|
| ai        | 2     | log(3 / (1+2)) = log(1)     | 0.000                |
| is        | 2     | log(3 / 3) = log(1)         | 0.000                |
| the       | 2     | log(3 / 3) = log(1)         | 0.000                |
| future    | 1     | log(3 / 2) ≈ 0.176          | 0.176                |
| world     | 2     | log(3 / 3) = log(1)         | 0.000                |
| data      | 1     | log(3 / 2) ≈ 0.176          | 0.176                |
| science   | 2     | log(3 / 3) = log(1)         | 0.000                |
| are       | 1     | log(3 / 2) ≈ 0.176          | 0.176                |
| powerful  | 1     | log(3 / 2) ≈ 0.176          | 0.176                |
| expanding | 1     | log(3 / 2) ≈ 0.176          | 0.176                |
| of        | 2     | log(3 / 3) = log(1)         | 0.000                |


🧠 Interpretation:
Words like "ai", "is", "the", "world" appear in most docs → IDF = 0, so not very informative.

Words like "future", "data", "expanding" are rarer → higher IDF → more informative.



💡 Interpretation
Words like "ai" and "future" have higher TF — they are more important in this document.

TF only looks at one document — it doesn’t care how rare/common a word is across documents (that’s what IDF does).

Unlike Bag of Words or TF-IDF (which are sparse and high-dimensional), Word2Vec captures semantic meaning.


# Transformers  : Eg : BERT,GPT2,ALBERT etc



⚙️ How Word2Vec Works
Word2Vec has two main architectures:

1. CBOW (Continuous Bag of Words)
Predicts the current word using surrounding context (before and after).

Example: Given ["the", "cat", "on", "the", "mat"], predict "sat".

2. Skip-Gram
Predicts the surrounding words using the current word.

Example: Given "sat", predict ["the", "cat", "on", "the", "mat"].

Skip-Gram performs better for rare words.
CBOW is faster and works well for frequent words.
