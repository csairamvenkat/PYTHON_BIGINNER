# Actual Arguments  are of 4 pasrts :
 Positional Argument
 Keyword Argument
 Default Argument
 Variable length Argument

```python
def person(name,age):  # formal args name and age
    print(name)
    print(age)
#person('sai',32)    

# psotional argument : Args are taken in the order of declaration

person(age=35,name='sai')  # keyword argument
```

```python
def person(name,age,city):  # formal args name and age
    print(name)
    print(age)

person(age=35,name='sai','Hyd')  
```

```python
def person(name,age,city):  # formal args name and age
    print(name)
    print(age)
    print(city)

person(age=35,name='sai',city='Hyd')  
```

```python
# Default Argument
def person(name,age=18):  # default argument and it shd be at last
    print(name)
    print(age)  # since we defaulted age to 18 if u dont pass default value is considered
    
    
    
person('sai')

person('ram',25)
```

# Variable Length
1. *args (Non-keyword arguments)
It allows you to pass a variable number of positional arguments.
The arguments are received as a tuple inside the function.

2. **kwargs (Keyword arguments)
It allows you to pass a variable number of keyword arguments.
The arguments are received as a dictionary inside the function.


```python
def add_numbers(*args):  # * before last arg
    return sum(args)

print(add_numbers(2, 3))          # Output: 5
print(add_numbers(1, 2, 3, 4, 5)) # Output: 15
```

```python
def add(x,*y): 
    c=x+y
    return c

add(5,6,7)
```

```python
def add(x,*y): 
    c=x
    for i in y:
        c=c+i
    print(c)

add(5,6,7)
```
