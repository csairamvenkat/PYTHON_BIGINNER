```python
df['Market_Status']=df['Market_Status'].str.replace(r'\W','',regex=True)
df['Market_Status']
```

```python
df['Market_Status'].unique()
```

```python
df['Market_Status'].replace("Err404", np.nan, inplace=True)
df['Market_Status'].replace("Corrupt", np.nan, inplace=True)
df['Market_Status'].replace("Unknown", np.nan, inplace=True)
df['Market_Status'].replace("Missing", np.nan, inplace=True)
df['Market_Status'].replace("NULL", np.nan, inplace=True)

df['Market_Status'].isnull().sum()
```

```python
df['Market_Status']=df['Market_Status'].fillna(df['Market_Status'].mode()[0])
df['Market_Status']
```
