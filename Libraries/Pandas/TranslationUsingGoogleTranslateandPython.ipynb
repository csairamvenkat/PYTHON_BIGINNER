```python
import pandas as pd
from googletrans import Translator

# Load your Excel file
file_path = 'your_excel_file.xlsx'  # Replace with your file path
df = pd.read_excel(file_path)

# Initialize the translator
translator = Translator()

# Column to translate
source_column = 'EnglishText'   # Replace with your source column name
target_column = 'SpanishText'   # Name of the new column

# Translate and store results
def translate_text(text):
    if pd.isna(text):
        return ''
    try:
        translated = translator.translate(text, src='en', dest='es')
        return translated.text
    except Exception as e:
        print(f"Translation error: {e}")
        return text  # fallback to original if error

df[target_column] = df[source_column].apply(translate_text)

# Save the result to a new Excel file
output_path = 'translated_output.xlsx'
df.to_excel(output_path, index=False)

print(f"Translation complete. File saved to: {output_path}")
```
