```python
# Nearest to traditional switch case

def switch_case_example(choice):
    match choice:
        case 1:
            return "Option 1 selected"
        case 2:
            return "Option 2 selected"
        case 3:
            return "Option 3 selected"
        case _:
            return "Invalid choice"

print(switch_case_example(2))  # Output: Option 2 selected
```

```python
# Break :
for i in range(5):
    if i == 3:
        break  # Loop stops when i is 3
    print(i)
```

```python
# Continue
for i in range(5):
    if i == 3:
        continue  # Skips the iteration when i is 3
    print(i)
```

```python
# Pass
for i in range(5):
    if i == 3:
        pass  # Does nothing
    print(i)
```

```python
def switch_case_example(choice):
    if choice == 1:
        return "Option 1 selected"
    elif choice == 2:
        return "Option 2 selected"
    elif choice == 3:
        return "Option 3 selected"
    else:
        return "Invalid choice"

print(switch_case_example(2))  # Output: Option 2 selected
```

```python
while True:
    num = int(input("Enter a number greater than 0: "))
    if num > 0:
        break  # Exit loop if condition is met
    print("Try again!")
```

```python
num = 0  # Initialize variable
while True:
    num += 1  # Perform operation (this executes at least once)
    print(f"Iteration: {num}")
    
    if num >= 5:  # Condition to stop loop
        break
```
