In Python, an iterator is an object that allows you to traverse through all the elements of a collection (like a list, tuple, or string), one at a time.
An iterator must implement two methods:

__iter__() → Returns the iterator object itself.

__next__() → Returns the next item in the sequence. Raises StopIteration when there are no more items.

An iterable is any object that can return an iterator (e.g., list, tuple, dictionary, set, string).



```python
my_list = [1, 2, 3]
my_iter = iter(my_list)  # Create an iterator object

print(next(my_iter))  # 1
print(next(my_iter))  # 2
print(next(my_iter))  # 3
print(next(my_iter))  # Raises StopIteration
```

If we use a nor mal function and use a return statement then if the index is passed it throws above error.

# Generator
To overcome this we use yield keyword instead of return which converts the normal Function to Generator.

# yield vs return
return ends the function and sends a value back once.

yield pauses the function and saves its state, resuming when next() is called again.

| Feature      | Iterator                                 | Generator                        |
| ------------ | ---------------------------------------- | -------------------------------- |
| Interface    | Implements `__iter__()` and `__next__()` | Uses `yield` instead of `return` |
| Creation     | Manually define class/methods            | Use a function with `yield`      |
| Memory Usage | Stores entire data                       | Yields one item at a time        |
| Use Case     | Custom iteration logic                   | Simple, lazy data production     |


```python
def count_up_to(n):
    count = 1
    while count <= n:
        yield count
        count += 1

# Usage
for number in count_up_to(3):
    print(number)
```

```python
def infinite_numbers():
    num = 0
    while True:
        yield num
        num += 1

gen = infinite_numbers()

for _ in range(5):
    print(next(gen))
```
