```python
class computer:
    
    def __init__(self):
        self.name = 'prakash'
        self.age = 37
    
    def update(self):  #lets create another object 
        self.age = 29 

c1 = computer()
c2 = computer()

print(id(c1))
print(id(c2))

c2.name = 'nit'
c2.age = 38

c2.update() 

print(c1.name, c1.age)
print(c2.name, c2.age)
```
