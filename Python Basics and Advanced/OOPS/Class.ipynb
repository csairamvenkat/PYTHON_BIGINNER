# Func inside class is called method



```python
class Wishing:
    def Hello(self):  # Method must include 'self' as the first parameter
        print("Hello")

obj = Wishing()  # Create an object of the class
obj.Hello()      # Call the method on the object
```

```python
class Wishing:
    @staticmethod
    def Hello():
        print("Hello")

# Call the method
Wishing.Hello()

# or via object
obj = Wishing()
obj.Hello()
```
