# Func inside class is called method



```python
class Wishing:
    def Hello(self):  # Method must include 'self' as the first parameter
        print("Hello")

obj = Wishing()  # Create an object of the class
obj.Hello()      # Call the method on the object
```

```python
class Wishing:
    @staticmethod
    def Hello():
        print("Hello")

# Call the method
Wishing.Hello()

# or via object
obj = Wishing()
obj.Hello()
```

1. self
self represents the instance of the class.

It allows access to the attributes and methods of the class in Python.

It must be the first parameter of any method in the class (including the constructor), although you don’t need to pass it explicitly when calling the method — Python does that for you.

2. Constructor: __init__
The __init__ method is the constructor in Python.

It’s automatically called when a new instance of the class is created.

You typically use it to initialize the object’s attributes.
