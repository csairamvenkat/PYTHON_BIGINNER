# Variable Length
1. *args (Non-keyword arguments)
It allows you to pass a variable number of positional arguments.
The arguments are received as a tuple inside the function.

2. **kwargs (Keyword arguments)
It allows you to pass a variable number of keyword arguments.
The arguments are received as a dictionary inside the function.


```python
def add_numbers(*args):  # * before last arg
    return sum(args)

print(add_numbers(2, 3))          # Output: 5
print(add_numbers(1, 2, 3, 4, 5)) # Output: 15
```

```python
def add(x,*y): 
    c=x+y
    return c

add(5,6,7)
```

```python
def add(x,*y): 
    c=x
    for i in y:
        c=c+i
    print(c)

add(5,6,7)
```

Local Variable is defined inside functions
Global  Variable is defined outside functions

Read Access – You can access a global variable inside a function without global.
Modify/Update – Use the global keyword inside a function to modify a global variable.

```python
a=10
def doSomething():
    b=15
    print("Local Variable :",b)
    print("Global Variable :",a)
doSomething() 
```

```python
a=20
def ABC():
    # global a
    a=15
    print('In function : ',a)
    a=9
ABC()
print('Out Function : ',a)
```

```python
a=20
def ABC():
    global a
    a=15
    print('In function : ',a)
    a=9
ABC()
print('Out Function : ',a)
```

Explanation:
Before the function call:

a is initialized to 20 (global scope).

Inside the function (ABC()):

global a allows you to modify the global a.

a = 15 changes the global variable to 15.

Prints "In function : 15".

Then, a = 9 updates the global variable to 9.

After the function call:

Global a is now 9.

Prints "Out Function : 9".

✅ Key Points:
The global keyword allows you to modify a global variable inside a function.

Without global, Python would treat a as a local variable, and the global value wouldn't change.

The final value of a remains 9 after the function call.

globals() and global can be used to access global variables


```python
# here global is used so global data is  updated .changes are global
x=10
def update_x():
    global x 
    x=x+5
    print(x)
update_x()
print(x)
    
```

```python
# here global is not used so global data is not updated .changes are scoped to function
x=10
def update_x():
    # global x 
    x=100
    x=x+5
    print(x)
update_x()
print(x)
    
```

```python
# here globals() is used so global data is updated 

x = 10

def update_x():
    globals()['x'] = globals()['x'] + 5  # Accessing global variable using globals()
    print(x)

update_x()
print(x)
```

# Anonymous function or  Lamda Functions

Func without name is called Anonymous function or  Lamda Functions


```python
f = lambda a: a * a  
result = f(5)  
print(result)
```

```python
# Filter
nums=[3,7,1,6,2,8,9,10.29,45]
# get even and odd  numbers in above list using lamda
even=list(filter(lambda e:e%2==0,nums))
odd=list(filter(lambda e:e%2!=0,nums))
print(even)
print(odd)
```

```python
# Map
# map() is used to apply a function to each item in an iterable (e.g., list, tuple) and return a new iterable.
nums=[3,7,1,6,2,8,9,10,29,45]
updated=list(map(lambda n:n*2,list(filter(lambda e:e%2==0,nums))))
print(updated)
```

```python
# reduce()
# # reduce() is used to apply a binary function (a function that takes two arguments) cumulatively to the elements of an iterable.
# It reduces the iterable to a single accumulated value.
from functools import reduce

nums = [3, 7, 1, 6, 2, 8, 9, 10, 29, 45]

# Step 1: Filter even numbers
filtered = list(filter(lambda e: e % 2 == 0, nums))  # [6, 2, 8, 10]

# Step 2: Double each filtered number
updated = list(map(lambda n: n * 2, filtered))  # [12, 4, 16, 20]

# Step 3: Reduce to sum all elements
sums = reduce(lambda a, b: a + b, updated)  # 12 + 4 + 16 + 20 = 52

print(sums)
```
