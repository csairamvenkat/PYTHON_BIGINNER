# LLM Tokenizer : It has BPE, Word Pience and Sentence Piece

BPE ENcoding :Byte Pair encoding 

Byte Pair Encoding (BPE) is a popular tokenization technique used in many large language models (LLMs), such as GPT-2 and GPT-3. It is a form of subword tokenization, which strikes a balance between character-level and word-level tokenization, making it well-suited for handling large vocabularies with rare and unknown words.

🔧 How BPE Works
Start with characters:

Each word is split into a sequence of characters (including special markers like Ġ for spaces in some implementations).

Example: "lower" → ["l", "o", "w", "e", "r"]

Build initial vocabulary:

Initialize the vocabulary with all unique characters in the training corpus.

Count adjacent pairs:

Find the most frequent pair of symbols (character or subword units) in the training corpus.

Merge pairs:

Replace all occurrences of that most frequent pair with a new token (e.g., merging "l" + "o" → "lo").

Add this new merged token to the vocabulary.

Repeat:

Repeat the merging process until the vocabulary reaches a pre-defined size or there are no more frequent pairs to merge.



# Word Piece :

🧩 WordPiece Tokenization (used in BERT and others)
WordPiece is another subword tokenization algorithm, similar to BPE, but with some key differences in how merges are chosen and how it's trained. It was originally developed by Google for speech recognition and later adapted for NLP, notably in BERT.

🔍 What is WordPiece?
WordPiece breaks words into subword units to handle rare and unknown words efficiently. Instead of only using full words in the vocabulary (as in traditional word-level tokenization), it builds a vocabulary of frequently occurring subword fragments.

⚙️ How WordPiece Works
1. Training Phase (vocabulary construction)
Goal: Find a vocabulary of subwords that minimizes the overall language model loss (usually likelihood or perplexity).

Start: With a base vocabulary of characters.

Build: New tokens are added iteratively based on which merge reduces the likelihood loss the most (not just frequency as in BPE).

Training is more data-driven and loss-based than BPE’s frequency-based merges.

2. Tokenization Phase (breaking input text)
Given a word, find the longest subword from the vocabulary that matches a prefix of the word.

Apply greedy longest-match-first strategy (also called maximum matching).

Continuation subwords are prefixed with ## to indicate they are part of a larger word.

 Comparison: WordPiece vs. BPE

| Feature             | WordPiece                            | BPE                       |
| ------------------- | ------------------------------------ | ------------------------- |
| Merge criterion     | Language model likelihood            | Frequency of symbol pairs |
| Merge granularity   | Greedy, longest subword match        | Greedy, based on pairs    |
| Continuation marker | `##` prefix (e.g., `##ing`)          | None in standard BPE      |
| OOV handling        | Breaks into subwords or uses `[UNK]` | Breaks into smaller units |
| Used in             | BERT, RoBERTa, ALBERT                | GPT-2, GPT-3, CLIP, etc.  |

✂️ SentencePiece: Tokenizer Used in T5, ALBERT, XLNet, etc.
SentencePiece is a language-independent tokenizer and detokenizer developed by Google. It is widely used in models like T5, ALBERT, XLNet, and mBERT. Unlike BPE or WordPiece, which often require language-specific pre-tokenization (like whitespace splitting), SentencePiece treats the input as a raw stream of text, including spaces.

🔧 Core Features of SentencePiece
| Feature              | SentencePiece                                          |
| -------------------- | ------------------------------------------------------ |
| Language-independent | No need for pre-tokenization (no whitespace splitting) |
| Input type           | Raw, untokenized text (Unicode)                        |
| Subword algorithms   | **Unigram Language Model** or **BPE**                  |
| Space handling       | Treats spaces as a regular symbol                      |
| Token format         | Uses `▁` (U+2581) to mark space boundaries             |
| Open-source?         | Yes – available via `sentencepiece` library            |


📘 Algorithms Used
1. Unigram Language Model (default)
Starts with a large set of candidate subwords.

Prunes the vocabulary to minimize the overall likelihood loss.

Picks the most likely subword sequence for each input.

Probabilistic model – supports multiple segmentations (useful for augmentation).

2. BPE (Byte Pair Encoding)
Optional. SentencePiece can train a BPE tokenizer if specified.

🧪 When to Use SentencePiece?
✅ If you:

Want language-independent tokenization

Are working with T5, ALBERT, mT5, mBART, XLNet

Need built-in detokenization

Want raw-text training without preprocessing


| Feature            | SentencePiece                | WordPiece                     | BPE                    |
| ------------------ | ---------------------------- | ----------------------------- | ---------------------- |
| Pre-tokenization   | ❌ Not needed                 | ✅ Needed (usually whitespace) | ✅ Needed               |
| Space handling     | Treated as special token `▁` | Special tokens or markers     | No standard approach   |
| Training algorithm | Unigram (default) / BPE      | Likelihood-based merges       | Frequency-based merges |
| OOV handling       | Great (subword fallback)     | `[UNK]` for unknowns          | Subword fallback       |
