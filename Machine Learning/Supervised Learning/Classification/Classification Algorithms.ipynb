# Logistic Regression Algorithm  : Its a proability Algorithm

Logistic Regression is a supervised machine learning algorithm used primarily for binary classification tasks (i.e., problems with two possible outcomes, such as yes/no, 0/1, true/false). Despite its name, it’s used for classification—not regression.

# Concept: :
It predicts the probability that a given input point belongs to a certain class. The output is a probability between 0 and 1, which is then mapped to a binary outcome.

 *)If dataset has outliers then we get miss classifications==>wrong predictions
 *)To adjust outliers we use probability functions so that outlier fits into S-Curve.
 *)The probability function used is called Sigmoid.
 *)S-Curve in Deep learning is called Sigmoid Activation.
 *) S-Curve ranges btw 0 to 1

# Logistic Regression Formula:
    At the heart of logistic regression is the logistic (sigmoid) function:

    Sigmoid(Z)=1/(1+e power -z)
   z=w0+w1x1+w2x2-----wnxn
  w are the weights (model parameters)
  x are the input features

   # How it Works:
    1)Linear Combination: Compute a weighted sum of the input features.
    2)Sigmoid Function: Apply the sigmoid function to squash the output into a probability range (0–1).
    3)Threshold: If probability > 0.5 → class 1; otherwise → class 0.

# ROC Curve and AUC 

The ROC curve (Receiver Operating Characteristic curve) is a key diagnostic tool in binary classification problems. It evaluates how well your classification model distinguishes between two classes across all classification thresholds.


# What is the ROC Curve?
The ROC curve is a plot of True Positive Rate (TPR) vs. False Positive Rate (FPR).

It helps visualize the trade-off between sensitivity and specificity.


Metric | Formula | Description
TPR (Recall) | TP / (TP + FN) | True Positives out of all actual positives
FPR | FP / (FP + TN) | False Positives out of all actual negatives

Interpreting ROC Curve:
A perfect model has a point at (0,1) — 0% false positives, 100% true positives.

A random model lies along the diagonal line (AUC = 0.5).

The closer the ROC curve is to the top-left, the better the model.



⭐ AUC (Area Under Curve):
AUC is a single number summary of the ROC curve.

Ranges from 0 to 1:

1.0 = perfect classification

0.5 = random guessing

< 0.5 = worse than random

```python
# Import necessary libraries
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt

# Step 1: Generate sample data
X, y = make_classification(n_samples=1000, n_features=20, 
                            n_classes=2, random_state=42)

# Step 2: Split data into train and test sets
x_train, x_test, y_train, y_test = train_test_split(X, y, 
                                                    test_size=0.3, 
                                                    random_state=42)

# Step 3: Train Logistic Regression model
best_model = LogisticRegression()
best_model.fit(x_train, y_train)

# Step 4: Get predicted probabilities for the positive class
y_prob = best_model.predict_proba(x_test)[:, 1]  # take probability of class 1

# Step 5: Calculate FPR, TPR, thresholds
fpr, tpr, thresholds = roc_curve(y_test, y_prob)
roc_auc = auc(fpr, tpr)

# Step 6: Plot ROC curve
plt.figure()
plt.plot(fpr, tpr, color='darkorange', lw=2, 
         label='ROC curve (AUC = {:.2f})'.format(roc_auc))
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')  # diagonal line
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic (ROC)')
plt.legend(loc="lower right")
plt.grid(True)
plt.show()
```
