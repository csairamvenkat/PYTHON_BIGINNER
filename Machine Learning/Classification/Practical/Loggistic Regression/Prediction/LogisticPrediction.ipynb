```python
import pandas as pd
import pickle

# Load the saved logistic regression model and scaler
with open('logistic_model.pkl', 'rb') as model_file:
    classifier = pickle.load(model_file)

with open('scaler.pkl', 'rb') as scaler_file:
    scaler = pickle.load(scaler_file)

# Load the new input CSV (update the path accordingly)
input_csv_path = r'C:\Users\sai.venkat\OneDrive\DataScience\Python\PYTHON_BIGINNER\Machine Learning\Classififcation\Practical\Loggistic Regression\Prediction\final1-PredictmodelGenerated.csv'
data = pd.read_csv(input_csv_path)
data = data.drop(columns=['User ID', 'Gender','Unnamed: 0'])

print(data)

# Extract only the features used for prediction (assuming columns at index 2 and 3 like before)
# X_new = data.iloc[:, [2]].values

X_new = data[['Age','EstimatedSalary']].values


# Apply the same scaler used during training
X_scaled = scaler.transform(X_new)

# Make predictions
predictions = classifier.predict(X_scaled)

# Add predictions to the original DataFrame
data['PredictedOutput'] = predictions

# Save the DataFrame with predictions to a new CSV
output_csv_path = r'C:\Users\sai.venkat\OneDrive\DataScience\Python\PYTHON_BIGINNER\Machine Learning\Classififcation\Practical\Loggistic Regression\Prediction\predicted_output.csv'
data.to_csv(output_csv_path, index=False)

print(f"Predictions saved to: {output_csv_path}")
```
